```
#@title 🔄 Alternative: Run test directly here
import asyncio
from atlas_agent import AtlasAgent, AtlasConfig

# Configure
URL = "https://example.com"  #@param {type:"string"}
TASK = "Verify the page loads and check the main heading"  #@param {type:"string"}

config = AtlasConfig(headless=True)

async def run_test():
    async with AtlasAgent(config) as agent:
        result = await agent.run_test(TASK, URL)
        print(f"\n{'='*50}")
        print(f"✅ Success: {result.success}")
        print(f"📝 Task: {result.task}")
        print(f"🔗 URL: {result.url}")
        print(f"🔄 Iterations: {result.iterations}")
        print(f"{'='*50}")
        print(f"\n📋 Result:\n{result.response}")

await run_test()
```
