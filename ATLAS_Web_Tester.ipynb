```
#@title 3️⃣ Run a Test (easiest method)
import nest_asyncio
nest_asyncio.apply()

from atlas_agent import AtlasAgent, AtlasConfig

#@markdown ### Enter your test details:
URL = "https://example.com"  #@param {type:"string"}
TASK = "Verify the page loads correctly and describe what you see"  #@param {type:"string"}

config = AtlasConfig(headless=True)

async def run_test():
    print(f"🔍 Testing: {URL}")
    print(f"📝 Task: {TASK}")
    print("\n⏳ Running test (this may take a minute)...\n")
    
    async with AtlasAgent(config) as agent:
        result = await agent.run_test(TASK, URL)
        
        print("="*60)
        if result.success:
            print("✅ TEST PASSED")
        else:
            print("❌ TEST FAILED")
        print("="*60)
        print(f"\n🔄 Iterations: {result.iterations}")
        print(f"\n📋 Result:\n\n{result.response}")
        if result.error:
            print(f"\n⚠️ Error: {result.error}")

await run_test()
```
